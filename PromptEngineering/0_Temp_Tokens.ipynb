# 🧠 LLM Lab: Tokens, Truncation, Temperature & Top-p (GitHub Models)

This notebook demonstrates:
- Large input & token pressure
- Output truncation using max_tokens
- Effect of temperature
- Effect of top_p

⚠️ Uses GitHub Models via chat.completions API


## 🔧 Setup

```python
# GitHub Models Configuration
import os
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv(dotenv_path="../.env")

client = OpenAI(
    base_url="https://models.github.ai/inference",
    api_key=os.getenv("GITHUB_TOKEN")
)

deployment_name = os.getenv("MODEL_DEPLOYMENT_NAME")

print("Model:", deployment_name)
print("✅ Setup complete")
```

```python
import os
from openai import AzureOpenAI

endpoint = "https://openaidemopvs.openai.azure.com/"
model_name = "gpt-4.1"
deployment = "gpt-4.1"

subscription_key = "asa"
api_version = "2024-12-01-preview"

client = AzureOpenAI(
    api_version=api_version,
    azure_endpoint=endpoint,
    api_key=subscription_key,
)
```

## 🔢 1. Large Input Test

```python
print("=== LARGE INPUT TEST ===")

large_text = "Artificial Intelligence is transforming the world. " * 300

print("Approx words:", len(large_text.split()))

response = client.chat.completions.create(
    model=deployment_name,
    messages=[{"role": "user", "content": large_text}],
    temperature=0.3,
    max_tokens=150
)

print("\nResponse:")
print(response.choices[0].message.content)
```

## ✂️ 2. Output Truncation

```python
print("=== OUTPUT TRUNCATION TEST ===")

prompt = "Explain Artificial Intelligence in detail."

for limit in [20, 50, 100]:
    response = client.chat.completions.create(
        model=deployment_name,
        messages=[{"role": "user", "content": prompt}],
        max_tokens=limit,
        temperature=0.3
    )

    print(f"\n--- max_tokens = {limit} ---")
    print(response.choices[0].message.content)
```

## 🎲 3. Temperature 

| Temperature            | Behavior                              |
| ---------------------- | ------------------------------------- |
| **Low (0.0 – 0.3)**    | More **accurate, consistent, similar every time** |
| **Medium (0.4 – 0.7)** | Balanced                              |
| **High (0.8 – 1.2)**   | More **creative, random, diverse**    |


```python
print("=== TEMPERATURE CONSISTENCY TEST ===")

prompt = "Explain Artificial Intelligence in a creative way."

for temp in [0.1, 1.0]:
    print(f"\n==============================")
    print(f"Temperature = {temp}")
    print(f"==============================")
    
    for i in range(3):  # run multiple times
        response = client.chat.completions.create(
            model=deployment_name,
            messages=[{"role": "user", "content": prompt}],
            temperature=temp,
            max_tokens=100
        )
        
        print(f"\nRun {i+1}:")
        print(response.choices[0].message.content)
```

## 📊 4. Top-p Effect

```python
print("=== TOP-P CONSISTENCY TEST ===")

prompt = "Explain Artificial Intelligence in a creative way."

for p in [0.5, 0.95]:
    print(f"\n==============================")
    print(f"Top-p = {p}")
    print(f"==============================")
    
    for i in range(3):
        response = client.chat.completions.create(
            model=deployment_name,
            messages=[{"role": "user", "content": prompt}],
            top_p=p,
            max_tokens=100
        )
        
        print(f"\nRun {i+1}:")
        print(response.choices[0].message.content)
```

## 🎯 Key Learnings
- Tokens act as budget
- max_tokens truncates output
- Temperature controls randomness
- Top-p controls diversity
