# Lab 04 — Multi-Agent Collaboration

In this lab we build **multiple AI agents that collaborate**.

Agents:

Research Agent → gathers information  
Writer Agent → writes explanation

Workflow:

User Question  
↓  
Research Agent  
↓  
Writer Agent  
↓  
Final Answer

## Setup GitHub Models

```python
import os
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()

client = OpenAI(
    api_key=os.getenv("GITHUB_TOKEN"),
    base_url="https://models.inference.ai.azure.com"
)
```

## Define Research Agent

```python
def research_agent(question):

    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role":"system","content":"You are a research assistant. Provide bullet point research notes."},
            {"role":"user","content":question}
        ]
    )

    return response.choices[0].message.content
```

## Define Writer Agent

```python
def writer_agent(research):

    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role":"system","content":"You are a writer. Convert research notes into a clear explanation."},
            {"role":"user","content":research}
        ]
    )

    return response.choices[0].message.content
```

## Run the Multi-Agent Workflow

```python
question = "Explain why AI agents are important."

research = research_agent(question)
print("Research notes:")
print(research)

final_answer = writer_agent(research)
print("\nFinal explanation:")
print(final_answer)
```
