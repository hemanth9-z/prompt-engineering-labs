# Lab 05 — Autonomous Research Agent

This lab builds an **autonomous agent** that:

1. Breaks a problem into steps
2. Solves each step
3. Produces a final report

Architecture:

Goal  
↓  
Planner  
↓  
Task Execution Loop  
↓  
Final Report

## Setup GitHub Models

```python
import os
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()

client = OpenAI(
    api_key=os.getenv("GITHUB_TOKEN"),
    base_url="https://models.inference.ai.azure.com"
)
```

## Step 1 — Planning Agent

```python
def planner(goal):

    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role":"system","content":"Break the goal into numbered research steps."},
            {"role":"user","content":goal}
        ]
    )

    return response.choices[0].message.content
```

## Step 2 — Task Executor

```python
def execute_task(task):

    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role":"system","content":"Execute the research step and summarize findings."},
            {"role":"user","content":task}
        ]
    )

    return response.choices[0].message.content
```

## Run Autonomous Agent

```python
goal = "Research the benefits of AI agents in healthcare"

plan = planner(goal)
print("Plan:")
print(plan)

steps = plan.split("\n")

results = []

for step in steps:
    if step.strip():
        result = execute_task(step)
        results.append(result)

print("\nResults:")
for r in results:
    print(r)
```
